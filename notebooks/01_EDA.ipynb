```python
import sys
from pathlib import Path

ROOT = Path('..').resolve()
sys.path.insert(0, str(ROOT))

import matplotlib.pyplot as plt
import pandas as pd
from IPython.display import Image, display, Markdown

FIGURES = ROOT / 'docs' / 'figures'
hourly = pd.read_csv(ROOT / 'data/processed/beijing/beijing_hourly_city.csv', parse_dates=['datetime'])
daily = pd.read_csv(ROOT / 'data/processed/beijing.csv', parse_dates=['date'])
aligned_bj = pd.read_csv(ROOT / 'data/processed/aligned/beijing.csv', parse_dates=['date'])
aligned_sh = pd.read_csv(ROOT / 'data/processed/aligned/shanghai.csv', parse_dates=['date'])
print(f'小时级行数: {len(hourly)}, 日级行数: {len(daily)}, 对齐子集行数: {len(aligned_bj)}')
```

## 图 1 — PM2.5 时序折线图

PM2.5 呈现明显季节性，采暖季冬季峰值显著高于夏季。这支持在模型中使用滞后特征和滚动统计特征。

```python
display(Image(filename=str(FIGURES / '01_pm25_timeseries.png')))
```

## 图 2 — PM2.5 分布直方图

分布呈右偏，存在长尾重污染日。模型评估应同时关注平均误差和极端 AQI 等级下的表现。

```python
display(Image(filename=str(FIGURES / '02_pm25_histogram.png')))
```

## 图 3 — 污染物浓度箱线图

颗粒物（PM2.5、PM10）为主要污染因子，气态污染物（SO2、NO2 等）中位数相对较低但偶有尖峰。

```python
display(Image(filename=str(FIGURES / '03_pollutant_boxplot.png')))
```

## 图 4 — 相关性热力图

PM2.5 与 PM10、NO2 正相关较强，与 O3 呈一定负相关，可为特征选择提供依据。

```python
display(Image(filename=str(FIGURES / '04_correlation_heatmap.png')))
```

## 图 5 — 缺失值热力图

短缺口插值降低了 PM2.5/PM10 的缺失率，同时保留了 CO/O3 等字段的真实长缺口，避免过度补值。

```python
display(Image(filename=str(FIGURES / '05_missing_heatmap.png')))
```

## 图 6 — 北京 vs 上海 PM2.5 对比
